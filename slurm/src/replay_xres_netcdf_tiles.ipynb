Cette exécution rejoue les 135 tuiles depuis leurs vecteurs de contrôle optimisés, sans relancer le 4DVar.

```python
import subprocess
import sys
from pathlib import Path

command = [
    sys.executable,
    '/home/flo/VarDyn/slurm/src/replay_xres_netcdf_tiles.py',
    '--config', '/home/flo/Workspace/UpperDyn/configs/config_VarDyn-QG_UpperDyn.py',
    '--config-eq', '/home/flo/Workspace/UpperDyn/configs/config_VarDyn-ID_UpperDyn_eq.py',
    '--source-controls', '/home/flo/Workspace/UpperDyn/controls/VarDyn-QG_UpperDyn_nadirs-minus-al_2024',
    '--source-scratch', '/home/flo/Workspace/UpperDyn/scratch/VarDyn-QG_UpperDyn_nadirs-minus-al_2024',
    '--output', '/home/flo/Workspace/UpperDyn/outputs/VarDyn-QG_UpperDyn_nadirs-minus-al_2024_netcdf',
    '--replay-controls', '/home/flo/Workspace/UpperDyn/controls/VarDyn-QG_UpperDyn_nadirs-minus-al_2024_netcdf_replay',
    '--timings', '/home/flo/Workspace/UpperDyn/diags/VarDyn-QG_UpperDyn_nadirs-minus-al_2024_zarr-vs-netcdf/netcdf_replay_timings.json',
    '--nprocs', '4',
    '--gpu-devices', '0,1',
    '--lat-min', '-12.5',
]
log_path = Path('/home/flo/Workspace/UpperDyn/runs/outs/VarDyn-QG_UpperDyn_nadirs-minus-al_2024-netcdf-replay.out')
log_path.parent.mkdir(parents=True, exist_ok=True)
with log_path.open('a') as log:
    subprocess.run(command, check=True, stdout=log, stderr=subprocess.STDOUT)
```
